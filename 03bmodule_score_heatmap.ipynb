```r
library(ComplexHeatmap)
library(circlize)
library(RColorBrewer)
```

```r
liwang_module <- read.table("/sc/arion/projects/roussp01a/liting/Olf/data/liwang_OSN_Goldstein_modulescore.txt",sep=',',header = T, row.names = 1, check.names = F)
pro_cells <- colnames(liwang_module)[grepl('RG|Immature|Newborn|IPC',colnames(liwang_module))]
liwang_module_prog <- liwang_module[,pro_cells]
```

```r
write.table(liwang_module_prog, file='./data/NC/f3b_liwang_module_prog.txt', sep='\t', quote = F)
```

```r
scale_liwang_module <- t(apply(liwang_module,1, scale))
colnames(scale_liwang_module) <- colnames(liwang_module)
scale_liwang_module_prog <- scale_liwang_module[,pro_cells]
```

```r
ComplexHeatmap::draw(Heatmap(
    scale_liwang_module,    
    cluster_rows = F,
    cluster_columns              = T,
    name='Module score',
    col= colorRamp2(seq(from=min(scale_liwang_module),to=max(scale_liwang_module),length=9),rev(brewer.pal(11, "RdBu")[c(-1,-11)])),
    row_gap = unit(c(2.5), "mm"),
    show_row_names               = T,
    show_column_names            = T,
    show_row_dend = T,

))
dev.print(pdf, file='/sc/arion/projects/roussp01a/liting/Olf/figures/liwang_OSN_Goldstein_Scaled_modulescore.pdf')
```
